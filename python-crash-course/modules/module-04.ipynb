## 1. Was ist eine Funktion?

Eine Funktion ist ein **wiederverwendbares Stück Code**, das eine bestimmte Aufgabe erledigt.  
Stell dir eine Funktion wie ein **Kochrezept** vor: Statt jeden Schritt immer wieder neu aufzuschreiben, rufst du einfach das Rezept auf.

**Warum Funktionen?**

* Sie **vermeiden Wiederholungen** im Code.
* Sie **strukturieren** Programme übersichtlich.
* Sie machen deinen Code **verständlicher und wartbarer**.

### 1.1 Eine Funktion definieren und aufrufen


```python
def begruessung():
    print("Hallo! Wie geht es dir?")
```

| Teil                      | Bedeutung                            |
| ------------------------- | ------------------------------------ |
| `def`                     | Start der Funktionsdefinition        |
| `begruessung()`           | Funktionsname + Klammern             |
| `:`                       | Einleitung des Funktionsblocks       |
| `print(...)` (eingerückt) | Der **Inhalt** (Körper) der Funktion |

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-info">

<strong class="notebook-title">INFO</strong>

Eine Funktion muss zuerst **definiert** werden. Erst danach kann sie **aufgerufen** werden.<br><br>

</div>


Aufrufen kannst du die Funktion, indem du ihren Namen mit Klammern schreibst:


```python
begruessung()
```

## 2. Funktionen mit Parametern

Funktionen können Werte entgegennehmen. Diese Werte heißen **Parameter**:


```python
def begruessung(name):
    print(f"Hallo {name}!")
```

```python
begruessung("Max")
begruessung("Lena")
```

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-info">

<strong class="notebook-title">INFO</strong>

`name` ist ein **Parameter** der Funktion. `"Max"` und `"Lena"` sind die **Argumente**, die du beim Aufruf übergibst.<br><br>

</div>


---

## 3. Funktionen mit Rückgabewert (`return`)

Oft soll eine Funktion ein **Ergebnis zurückgeben**, mit dem du weiterarbeiten kannst. Dafür gibt es `return`:


```python
def quadrat(zahl):
    return zahl * zahl

ergebnis = quadrat(5)
print(ergebnis)
```

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-hint">

<strong class="notebook-title">TIPP</strong>

**`print` oder `return`?**

* `print(...)` **zeigt** nur etwas auf dem Bildschirm an.
* `return ...` **gibt einen Wert zurück**, mit dem du weiterrechnen kannst (z. B. `ergebnis = quadrat(5)`).

Für die folgenden Aufgaben ist das wichtig: Damit die Test-Zellen dein Ergebnis prüfen können, müssen die Funktionen ihr Ergebnis mit `return` **zurückgeben**.<br><br>

</div>
