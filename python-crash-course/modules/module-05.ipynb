## 1. Listen

### 1.1 Was ist eine Liste?

Eine **Liste** ist eine Sammlung mehrerer Werte in einer festen Reihenfolge.  
Man kann sie sich wie ein Regal mit mehreren nummerierten Fächern vorstellen.


```python
namen = ["Lena", "Max", "Tom"]
zahlen = [1, 2, 3, 4, 5]
```

Auf einzelne Einträge greifst du über ihren **Index** (die Position) zu:


```python
print(namen[0])  # Lena
print(namen[1])  # Max
```

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-info">

<strong class="notebook-title">INFO</strong>

Der erste Eintrag hat immer den Index `0`, der zweite `1` usw.<br><br>

</div>


### 1.2 Listen verändern

Du kannst Einträge **hinzufügen**, **entfernen** und **ersetzen**:


```python
namen.append("Nina")   # hinzufügen (ans Ende)
print(namen)

namen.remove("Max")    # entfernen (nach Wert)
print(namen)

namen[0] = "Ben"       # ersetzen (nach Index)
print(namen)
```

## 2. Tupel

Ein **Tupel** ist fast wie eine Liste, aber **unveränderlich** („read-only“).  
Du erkennst Tupel an den **runden Klammern** `()`:


```python
punkte = (10, 20, 30)
```

Tupel eignen sich gut für feste Daten, z. B. Koordinaten:


```python
koordinaten = (13.4, 52.5)
print(f"Startpunkt: x = {koordinaten[0]} | y = {koordinaten[1]}")
```

| Typ   | Klammern | Veränderbar |
| ----- | -------- | ----------- |
| Liste | `[]`     | ✅ Ja       |
| Tupel | `()`     | ❌ Nein     |

## 3. Schleifen

Mit einer **Schleife** wird Code wiederholt ausgeführt.

### 3.1 `for`-Schleife

Perfekt, um über eine Liste zu laufen:


```python
for name in namen:
    print("Hallo", name)
```

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-info">

<strong class="notebook-title">INFO</strong>

Eine `for`-Schleife nutzt du in der Regel, wenn die **Anzahl der Durchläufe bekannt** ist.<br>
Mit `range(start, stop)` zählst du durch einen Zahlenbereich (`stop` ist ausgeschlossen).<br><br>

</div>


```python
for num in range(0, 5):
    print(f"Zahl: {num}")
```

### 3.2 `while`-Schleife

Eine `while`-Schleife läuft **so lange**, wie eine Bedingung **wahr** ist:


```python
x = 0

while x < 5:
    print("x ist:", x)
    x += 1
```
