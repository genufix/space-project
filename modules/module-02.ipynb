## 1. Was ist ein Datentyp?

Ein **Datentyp** beschreibt, welche Art von Information in einem Wert steckt.<br>
Python erkennt den Datentyp automatisch – je nachdem, was du schreibst.

### 1.1 Grundlegende Datentypen in Python

| Datentyp | Bedeutung           | Beispiel         |
| -------- | ------------------- | ---------------- |
| `int`    | Ganze Zahl          | `5`, `-100`, `0` |
| `float`  | Kommazahl           | `3.14`, `-0.1`   |
| `str`    | Zeichenkette (Text) | `"Hallo"`        |
| `bool`   | Wahr oder falsch    | `True`, `False`  |

### 1.2 Datentyp herausfinden

Mit der Funktion `type()` kannst du dir den Datentyp eines Werts anzeigen lassen:


````python
print(type(5))
print(type(3.14))
print(type("Hallo"))
print(type(True))
````

---

## 2. Was ist eine Variable?

Eine Variable ist wie ein **Behälter mit einem Namen**, in dem du einen Wert speicherst.  
Du vergibst den Namen einmal und kannst den Wert danach immer wieder verwenden.


````python
alter = 25
name = "Max"
pi = 3.14
ist_azubi = False
````

Du kannst den Inhalt jederzeit weiterverwenden:


````python
print(alter)
print(name)
print(pi)
print(ist_azubi)
````

### 2.1 Regeln für Variablennamen

#### ✅ Erlaubt
* Buchstaben (a–z, A–Z)
* Ziffern (aber **nicht** am Anfang)
* Unterstriche `_`

#### ⛔ Nicht erlaubt
* Sonderzeichen wie `!`, `@`, `#`
* Leerzeichen
* Python-Schlüsselwörter wie `if`, `class`, `while`

### 2.2 Beispiele

```py
name = "Max"          # ✅ gültig
alter_2024 = 25       # ✅ gültig
2name = "Tom"         # ❌ Fehler: beginnt mit einer Zahl
mein name = "Lucas"   # ❌ Fehler: enthält ein Leerzeichen
```

## 3. String-Manipulation

**Strings** (Zeichenketten) kannst du zusammenfügen (*konkatenieren*) oder durch Formatierung besser lesbar machen.

### 3.1 Möglichkeit 1: Mit `+` verketten


````python
vorname = "Max"
nachname = "Mustermann"
````

````python
name = vorname + " " + nachname
print(name)
````

### 3.2 Möglichkeit 2: `f-Strings` verwenden (empfohlen)


````python
name = f"{vorname} {nachname}"
print(name)
````

### 3.3 Möglichkeit 3: `print()` mit Kommas


````python
print(vorname, nachname)
````

## 4. Mathematische Operationen

In Python kannst du ganz einfach rechnen. Hier eine Übersicht:

| Operation | Beispiel  | Bedeutung                  |
| --------- | --------- | -------------------------- |
| `+`       | `2 + 3`   | Addition                   |
| `-`       | `5 - 1`   | Subtraktion                |
| `*`       | `4 * 2`   | Multiplikation             |
| `/`       | `10 / 2`  | Division (Ergebnis: float) |
| `//`      | `10 // 3` | Ganzzahldivision           |
| `%`       | `10 % 3`  | Modulo (Rest)              |
| `**`      | `2 ** 3`  | Potenz (2 hoch 3 = 8)      |


````python
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(2 ** 3)
````
