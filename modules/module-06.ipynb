## 1. Was ist ein Dictionary?

Statt Werte nur über eine Position (wie bei Listen) anzusprechen, gibst du jedem Wert einen **Schlüssel** (`key`):


```python
person = {
    "name": "Max",
    "alter": 25,
    "ausgeschlafen": True,
}
person
```

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-info">

<strong class="notebook-title">INFO</strong>

Du erkennst **Dictionaries** an den **geschweiften Klammern** `{}` und an den `"schlüssel": wert`-Paaren.<br><br>

</div>


---

## 2. Zugriff auf Werte

Einzelne Werte holst du über ihren Schlüssel:


```python
print(person["name"])
print(person["alter"])
```

---

## 3. Werte ändern, hinzufügen & löschen


```python
person["alter"] = 26            # vorhandenen Wert ändern
person["stadt"] = "Kirchheim"   # neuen Schlüssel hinzufügen
del person["ausgeschlafen"]     # Schlüssel löschen
print(person)
```

---

## 4. Über Dictionaries iterieren

Mit einer `for`-Schleife kannst du über ein Dictionary laufen:


```python
for key in person:
    print(key, ":", person[key])
```

Noch lesbarer geht es mit `.items()`, das Schlüssel **und** Wert liefert:


```python
for key, value in person.items():
    print(f"{key}: {value}")
```
