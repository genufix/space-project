## 1. Was ist eine Kontrollstruktur?

Kontrollstrukturen steuern, **welcher Teil deines Codes** ausgeführt wird – abhängig von einer Bedingung.  
In Python wird dafür die `if`-Anweisung verwendet:


````python
ist_praktikant = True

if ist_praktikant:
    print("Du bist heute Praktikant bei genua!")
    print("Herzlich Willkommen. :)")
````

<strong class="notebook-title">INFO</strong>

Ist die Bedingung erfüllt (`True`), wird der **eingerückte** Code darunter ausgeführt.<br><br>

## 2. Bedingungen mit `if`, `elif` & `else`

Oft gibt es mehr als zwei Fälle. Mit `elif` („else if“) prüfst du weitere Bedingungen, mit `else` fängst du alle übrigen Fälle ab:


````python
note = 2

if note == 1:
    print("Note: Sehr gut")
elif note == 2:
    print("Note: Gut")
elif note == 3:
    print("Note: Befriedigend")
elif note == 4:
    print("Note: Ausreichend")
elif note == 5:
    print("Note: Mangelhaft")
elif note == 6:
    print("Note: Ungenügend")
else:
    print("Du hast am Test nicht teilgenommen!")
````

## 3. Vergleichsoperatoren

Mit Vergleichsoperatoren vergleichst du zwei Werte. Das Ergebnis ist immer `True` oder `False`.

| Operator | Bedeutung           | Beispiel  |
| -------- | ------------------- | --------- |
| `==`     | Gleich              | `x == 5`  |
| `!=`     | Ungleich            | `x != 10` |
| `>`      | Größer als          | `x > 3`   |
| `<`      | Kleiner als         | `x < 7`   |
| `>=`     | Größer oder gleich  | `x >= 18` |
| `<=`     | Kleiner oder gleich | `x <= 0`  |


---

## 4. Logische Operatoren

Mit logischen Operatoren kombinierst du mehrere Bedingungen.

| Operator | Bedeutung                     | Beispiel            |
| -------- | ----------------------------- | ------------------- |
| `and`    | Und – beide müssen wahr sein  | `x > 0 and x < 10`  |
| `or`     | Oder – eine Bedingung reicht  | `x < 0 or x > 100`  |
| `not`    | Nicht – kehrt den Wert um     | `not ausgeschlafen` |

<link rel="stylesheet" href="../assets/notebook.css">
<div class="notebook-box notebook-info">

<strong class="notebook-title">INFO</strong>

Mehr dazu findest du im [Python Cheatsheet](https://www.pythoncheatsheet.org/cheatsheet/control-flow).<br><br>

</div>


````python
alter = 20
hat_ausweis = True

# and: beide Bedingungen müssen wahr sein
if alter >= 18 and hat_ausweis:
    print("Zutritt erlaubt")

# or: eine Bedingung reicht (entweder ... oder)
temperatur = 35
if temperatur < 0 or temperatur > 30:
    print("Extreme Temperatur!")

# not: kehrt den Wahrheitswert um
ausgeschlafen = False
if not ausgeschlafen:
    print("Bitte geh heute früher ins Bett!")
````
