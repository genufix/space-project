## 1. Was ist OOP?

**Objektorientierte Programmierung (OOP)** ist ein Programmieransatz, bei dem man **eigene Datentypen (Klassen)** definiert, um **Dinge aus der echten Welt nachzubilden**.  
Statt mit losen Variablen und Funktionen zu arbeiten, **bündelt man Daten und Verhalten logisch zusammen**.

## 2. Was ist eine Klasse?

Eine **Klasse** ist wie ein **Bauplan** für ein Objekt. Sie beschreibt:

* **welche Eigenschaften** das Objekt hat (z. B. `name`, `alter`)
* **was es tun kann** (z. B. `vorstellen()`)

Mit einer Klasse kannst du **beliebig viele Objekte** erzeugen, die nach diesem Bauplan funktionieren.


---

## 3. Beispiel: Klasse `Person`


```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def vorstellen(self):
        print(f"Hallo, ich heiße {self.name} und bin {self.alter} Jahre alt.")
```

* `class Person`: definiert eine neue Klasse mit dem Namen `Person`.
* `__init__` ist der **Konstruktor**. Er wird **automatisch** aufgerufen, wenn ein neues Objekt erstellt wird.
* `self` steht immer für **das aktuelle Objekt**.
* `self.name = name`: speichert den Namen im Objekt.
* `vorstellen()` ist eine **Methode** – eine Funktion, die zum Objekt gehört.

```python
person1 = Person("Max", 25)
person1.vorstellen()

person2 = Person("Lena", 22)
person2.vorstellen()
```

## 4. Warum sind Klassen so praktisch?

| Vorteil                  | Beschreibung                                                          |
| ------------------------ | -------------------------------------------------------------------- |
| **Struktur & Ordnung**   | Zusammengehörige Daten und Verhalten werden übersichtlich gruppiert. |
| **Wiederverwendbarkeit** | Objekte lassen sich mehrfach und in verschiedenen Kontexten nutzen.  |
| **Erweiterbarkeit**      | Neue Funktionen lassen sich leicht in die Klasse einbauen.           |
| **Realitätsnähe**        | OOP orientiert sich an Dingen aus der echten Welt.                   |
